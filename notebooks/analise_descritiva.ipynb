# **1. Estatística Descritiva**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1 - Carregar arquivo limpo
df = pd.read_csv('wdbc_limpo.csv')

# Remover a coluna 'id' para não distorcer as estatísticas
if 'id' in df.columns:
    df = df.drop('id', axis=1)

print("Dimensões do Dataset:", df.shape)

# Separando as variáveis (X) para os cálculos gerais
X = df.drop('diagnosis', axis=1)

# 2 - Estatísticas Descritivas Gerais
desc_stats = X.describe().T
desc_stats['variance'] = X.var()
desc_stats['skewness'] = X.skew()
desc_stats['kurtosis'] = X.kurtosis()

print("\n--- ESTATÍSTICAS DESCRITIVAS GERAIS ---")
display(desc_stats.round(4))

# 3 - Estatísticas Estratificadas por Classe
df_benigno = df[df['diagnosis'] == 0].drop('diagnosis', axis=1)
df_maligno = df[df['diagnosis'] == 1].drop('diagnosis', axis=1)

# Benignos
desc_benigno = df_benigno.describe().T
desc_benigno['variance'] = df_benigno.var()
desc_benigno['skewness'] = df_benigno.skew()
desc_benigno['kurtosis'] = df_benigno.kurtosis()

# Malignos
desc_maligno = df_maligno.describe().T
desc_maligno['variance'] = df_maligno.var()
desc_maligno['skewness'] = df_maligno.skew()
desc_maligno['kurtosis'] = df_maligno.kurtosis()

print("\n---> ESTATÍSTICAS: CLASSE BENIGNA (0)")
display(desc_benigno.round(4))

print("\n---> ESTATÍSTICAS: CLASSE MALIGNA (1)")
display(desc_maligno.round(4))
```

# **2. Distribuição e Separação Visual**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1 - Selecionando apenas as variáveis que representam as médias (10 variáveis)
features_mean = [col for col in df.columns if '_mean' in col]

# Mapeando os nomes para a legenda ficar mais bonita
df['diagnosis_label'] = df['diagnosis'].map({0: 'Benigno', 1: 'Maligno'})

# 2 - Configurando o grid de gráficos (KDE e Boxplot lado a lado)
for feature in features_mean:
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))

    # Gráfico 1: Curva de Densidade (KDE)
    sns.kdeplot(data=df, x=feature, hue='diagnosis_label', fill=True,
                palette={'Benigno': 'blue', 'Maligno': 'red'}, ax=axes[0])
    axes[0].set_title(f'Distribuição (KDE) - {feature}')

    # Gráfico 2: Boxplot (Corrigido para não dar FutureWarning)
    sns.boxplot(data=df, x='diagnosis_label', y=feature, hue='diagnosis_label',
                palette={'Benigno': 'blue', 'Maligno': 'red'}, ax=axes[1], legend=False)
    axes[1].set_title(f'Boxplot e Outliers - {feature}')
    axes[1].set_xlabel('')

    plt.tight_layout()
    plt.show()

# Removendo a coluna de rótulo para não atrapalhar os próximos passos
df = df.drop('diagnosis_label', axis=1)
```

# **3. Heatmap de Multicolinearidade**

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1 - Isolando as variáveis independentes para a matriz
X = df.drop('diagnosis', axis=1)

# 2 - Calculando a matriz de correlação (Pearson)
corr_matrix = X.corr()

# 3 - Configurando a visualização do Heatmap
plt.figure(figsize=(20, 16))

# Criando uma máscara para esconder a parte superior do gráfico (que é um espelho da inferior)
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Plotando o Heatmap
sns.heatmap(corr_matrix, mask=mask, annot=False, cmap='coolwarm',
            vmin=-1, vmax=1, center=0, linewidths=0.5)

plt.title('Matriz de Correlação - Diagnóstico de Multicolinearidade', fontsize=18)
plt.tight_layout()
plt.show()

# 4 - Extraindo automaticamente os pares com correlação perigosa (> 0.90)
print("\n---> PARES ALTAMENTE CORRELACIONADOS (> 0.90)")

# Pegando apenas o triângulo superior da matriz para não duplicar os pares
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Encontrando os cruzamentos com valor absoluto maior que 0.90
high_corr = [(upper_tri.index[x], upper_tri.columns[y], upper_tri.iloc[x, y])
             for x, y in zip(*np.where(upper_tri.abs() > 0.90))]

# Exibindo os resultados ordenados do maior para o menor
for var1, var2, valor in sorted(high_corr, key=lambda x: abs(x[2]), reverse=True):
    print(f"-> {var1} e {var2}: {valor:.4f}")
```

# **4. Pairplots das Melhores Variáveis**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1 - Selecionar um subconjunto estratégico de variáveis (sobreviventes da multicolinearidade)
# Escolhemos uma medida de tamanho (raio), uma de textura, uma de suavidade e uma de forma (concavidade).
selected_features = ['radius_mean', 'texture_mean', 'smoothness_mean', 'concavity_mean']

# 2 - Criar um DataFrame temporário focado apenas nestas variáveis para o gráfico
df_pairplot = df[selected_features].copy()

# Adicionar os rótulos de forma legível
df_pairplot['Diagnóstico'] = df['diagnosis'].map({0: 'Benigno', 1: 'Maligno'})

# 3 - Gerar a matriz de dispersão (Pairplot)
sns.pairplot(df_pairplot, hue='Diagnóstico',
             palette={'Benigno': '#1f77b4', 'Maligno': '#d62728'},
             markers=["o", "s"],
             plot_kws={'alpha': 0.7, 'edgecolor': 'k'})

plt.suptitle('Cruzamento das Variáveis com Maior Poder Discriminante', y=1.02, fontsize=16)
plt.show()
```
